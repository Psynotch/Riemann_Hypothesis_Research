```python

import mpmath as mp
import numpy as np
import matplotlib.pyplot as plt

mp.mp.dps = 40

N = 100

zeros = [mp.zetazero(n) for n in range(1, N + 1)]

real_parts = [float(mp.re(z)) for z in zeros]
imag_parts = [float(mp.im(z)) for z in zeros]

print("First 10 non-trivial zeros:\n")

for i, z in enumerate(zeros[:10], 1):
    print(i, mp.nstr(z, 25))

plt.figure(figsize=(8, 6))
plt.scatter(real_parts, imag_parts, s=15)
plt.axvline(0.5, linestyle="--", label="Critical line")
plt.xlabel("Real part")
plt.ylabel("Imaginary part")
plt.title("First 100 Riemann Zeta Zeros")
plt.legend()
plt.grid()
plt.show()
```

```python
#Experiment B - Investigating the spacing between zeros

gaps = np.diff(imag_parts)

print("First 10 zero spacings:")
for i, gap in enumerate(gaps[:10], 1):
    print(f"Gap {i}: {gap:.10f}")

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(gaps)+1), gaps, marker="o")
plt.xlabel("Zero interval")
plt.ylabel("Spacing")
plt.title("Spacing Between Consecutive Zeros")
plt.grid()
plt.show()

print("Mean spacing:", np.mean(gaps))
print("Minimum spacing:", np.min(gaps))
print("Maximum spacing:", np.max(gaps))
```

```python
#Normalizing Spacing

normalized_gaps = []

for i in range(len(gaps)):
    T = (imag_parts[i] + imag_parts[i+1]) / 2
    mean_gap = 2 * np.pi / np.log(T / (2 * np.pi))
    normalized_gaps.append(gaps[i] / mean_gap)

plt.figure(figsize=(9, 5))
plt.hist(normalized_gaps, bins=15, density=True)
plt.xlabel("Normalized zero spacing")
plt.ylabel("Density")
plt.title("Normalized Riemann Zero Spacing")
plt.grid()
plt.show()
```

```python
#Experiment C - Test the zero-counting formula

def zero_count_approx(T):
    return (
        T/(2*np.pi)*np.log(T/(2*np.pi))
        - T/(2*np.pi)
        + 7/8
    )

heights = imag_parts

actual_counts = np.arange(1, N+1)
predicted_counts = [
    zero_count_approx(T) for T in heights
]

plt.figure(figsize=(9, 5))
plt.plot(heights, actual_counts, label="Actual count")
plt.plot(heights, predicted_counts, "--",
         label="Riemann-von Mangoldt approximation")
plt.xlabel("Height T")
plt.ylabel("Number of zeros")
plt.legend()
plt.title("Zero Counting Function")
plt.grid()
plt.show()

errors = actual_counts - predicted_counts

plt.figure(figsize=(9, 5))
plt.plot(heights, errors)
plt.xlabel("Height T")
plt.ylabel("Counting error")
plt.title("Error in Zero Counting Approximation")
plt.grid()
plt.show()
```

```python
#Experiment D - Function away from critical line

import matplotlib.pyplot as plt
import numpy as np
import mpmath as mp

mp.mp.dps = 30

t_values = np.linspace(10, 40, 500)

for sigma in [0.3, 0.5, 0.7]:
    magnitudes = []

    for t in t_values:
        value = mp.zeta(
            mp.mpc(sigma, float(t))
        )
        magnitudes.append(float(abs(value)))

    plt.plot(t_values, magnitudes,
             label=f"Re(s) = {sigma}")

plt.yscale("log")
plt.xlabel("Imaginary part (t)")
plt.ylabel("|ζ(σ + it)|")
plt.title("Zeta Function Away From and On the Critical Line")
plt.legend()
plt.grid()
plt.show()
```
