```python
from flint import acb, arb, ctx
from fractions import Fraction

ctx.prec = 192

print("Session 05 initialized")
print("Precision:", ctx.prec, "bits")
```

```python
#Hardy's z function
def rational_ball(q):
    return arb(f"{q.numerator}/{q.denominator}")


def hardy_z(t_fraction):
    """
    Rigorous ball enclosure of Hardy's Z(t)
    at an exact rational value of t.
    """

    t = rational_ball(t_fraction)

    # s = 1/2 + it
    s = acb(arb(1) / 2, t)

    # Gamma argument = 1/4 + it/2
    gamma_arg = acb(arb(1) / 4, t / 2)

    # Riemann-Siegel theta function
    theta = (
        gamma_arg.lgamma().imag
        - (t / 2) * arb.pi().log()
    )

    # exp(i * theta)
    phase = acb(0, theta).exp()

    # Hardy Z(t)
    result = phase * s.zeta()

    return result.real, result.imag


print("Hardy Z-function defined.")
```

```python
#13-intervals
zero_centers = [
    "14.13472514173469379045725",
    "21.02203963877155499262848",
    "25.01085758014568876321379",
    "30.4248761258595132103119",
    "32.93506158773918969066237",
    "37.58617815882567125721776",
    "40.91871901214749518739813",
    "43.32707328091499951949612",
    "48.00515088116715972794247",
    "49.77383247767230218191678",
    "52.9703214777144606441473",
    "56.44624769706339480436776",
    "59.34704400260235307965365"
]

radius = Fraction(1, 100)

intervals = []

for center_text in zero_centers:
    center = Fraction(center_text)

    intervals.append((
        center - radius,
        center + radius
    ))

# Check all intervals are inside the verified height range
assert all(a > 10 and b < 60 for a, b in intervals)

# Check they are mutually disjoint
assert all(
    intervals[i][1] < intervals[i+1][0]
    for i in range(len(intervals)-1)
)

print("Intervals created:", len(intervals))
print("All intervals are disjoint:", True)
print("All intervals are inside (10, 60):", True)
```

```python
#Rigorous Sign-Change verification
def certified_sign(value):
    sign = value.sgn()

    if sign == arb(1):
        return "+"

    if sign == arb(-1):
        return "-"

    return "UNKNOWN"


results = []

print("HARDY Z SIGN-CHANGE VERIFICATION")
print("--------------------------------")

for i, (left, right) in enumerate(intervals, start=1):

    z_left, im_left = hardy_z(left)
    z_right, im_right = hardy_z(right)

    sign_left = certified_sign(z_left)
    sign_right = certified_sign(z_right)

    real_check = (
        im_left.contains(0)
        and im_right.contains(0)
    )

    sign_change = (
        sign_left != "UNKNOWN"
        and sign_right != "UNKNOWN"
        and sign_left != sign_right
    )

    success = real_check and sign_change

    results.append({
        "zero": i,
        "left_sign": sign_left,
        "right_sign": sign_right,
        "real_check": real_check,
        "sign_change": sign_change,
        "success": success
    })

    print(
        f"Zero {i:02d}: "
        f"Z(left)={sign_left}, "
        f"Z(right)={sign_right}, "
        f"REAL={real_check}, "
        f"PASS={success}"
    )

verified = sum(r["success"] for r in results)

print("\nVERIFIED SIGN CHANGES:", verified)
print("TOTAL INTERVALS:", len(intervals))
```
